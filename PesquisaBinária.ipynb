```python
def pesquisa_binaria(lista, item):
  baixo = 0  # Inicializa 'baixo' com 0, o índice do primeiro elemento da lista.
  alto = len(lista) - 1 # Inicializa 'alto' com o índice do último elemento da lista.
  while baixo <= alto: # Continua o loop enquanto a sublista a ser pesquisada for válida.
    meio = (baixo + alto) // 2 # Calcula o índice do meio da sublista atual.
    chute = lista[int(meio)] # Obtém o valor no índice do meio.
    if chute == item: # Se o valor do meio é o item procurado...
      return meio # ...retorna o índice do item.
    if chute > item: # Se o valor do meio é maior que o item procurado...
      alto = int(meio) - 1 # ...descarta a metade superior da sublista.
    else: # Se o valor do meio é menor que o item procurado...
      baixo = int(meio) + 1 # ...descarta a metade inferior da sublista.
  return None # Se o loop termina e o item não foi encontrado, retorna None.

minha_lista = [1, 3, 5, 7, 9, 11, 13, 15] # Cria uma lista de exemplo para testar a função.

print("posição:", pesquisa_binaria(minha_lista, 11)) # Chama a função para buscar '5' e imprime o resultado.
print("posição:", pesquisa_binaria(minha_lista, 3)) # Chama a função para buscar '2' e imprime o resultado.
```
